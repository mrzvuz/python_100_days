## Part 1: For Loop Exercises

A `for` loop repeats code for every item in a collection.

### Exercise 1: Print Names

Create a list of 5 names. Use a `for` loop to print each name.

Expected idea:

```python
Hello Adam
Hello Sara
...
```

````python
# Your answer for loop exercise 1

names = ["Alice", "Bob", "Charlie", "David", "Eve"]
for name in names:
    print(f"Hello, {name}! Welcome to the Python course.")
````

### Exercise 2: Total Prices

Create a list of prices: `10`, `25`, `40`, `5`.

Use a `for` loop to calculate the total price and print the final total.

````python
# Your answer for loop exercise 2
prices = [10,25,40, 5]

total_price = 0
for price in prices:
    total_price += price
print(f"Total price: ${total_price:.2f}")
````

### Exercise 3: Product Stock Value

You have products with 3 values each:

1. name
2. price
3. stock

Use a `for` loop to print each product's total stock value.

`total stock value = price * stock`

````python
# Your answer for loop exercise 3
products = [
    ["Laptop", 1200, 5],
    ["Mouse", 25, 50],
    ["Notebook", 4, 100],
]

for product in products:
    name, price, quantity = product
    total_value = price * quantity
    print(f"Product: {name}, Total Value: ${total_value:.2f}")
````

## Part 2: While Loop Exercises

A `while` loop repeats code while a condition is `True`.

Use a `while` loop when you want to continue until something changes.

### Exercise 4: Count From 1 to 5

Create a variable called `count` and set it to `1`.

Use a `while` loop to print numbers from `1` to `5`.

````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

### Exercise 5: Countdown

Create a variable called `number` and set it to `5`.

Use a `while` loop to print numbers from `5` down to `1`.

````python
# Your answer for while loop exercise 2
number = 5

while number >= 1:
    print(number)
    number -= 1
````

### Exercise 6: Savings Goal

You start with `balance = 0`.

Each week, you save `50`.

Use a `while` loop to keep saving until the balance reaches at least `200`.

Print the balance each week.

````python
balance = 0
weekly_saving = 50
week = 1

while balance < 200:
    balance = balance + weekly_saving
    print("Week", week, "balance:", balance)
    week = week + 1
````

## Part 3: Function Exercises

A function is reusable code. It can receive inputs called parameters.

### Exercise 7: Full Name Function

Create a function called `create_full_name` with 3 parameters:

1. `first_name`
2. `middle_name`
3. `last_name`

Return the full name as one string.

### Exercise 8: Order Total Function

Create a function called `calculate_order_total` with 4 parameters:

1. `product_name`
2. `quantity`
3. `price_each`
4. `tax_rate`

Return the final total after tax.

`subtotal = quantity * price_each`

`total = subtotal + subtotal * tax_rate`

````python
# Your answer for function exercise 2
def calculate_order_total(product_name, quantity, price_each, tax_rate):
    subtotal = quantity * price_each
    total = subtotal + subtotal * tax_rate
    return f"Product Name: {product_name} | Quantity: {quantity} | Price Per Item: {price_each:.2f} | Total Price: ${total:.2f}"

print(calculate_order_total("Laptop", 5, 1200, 0.08))
````

### Exercise 9: Student Average Function

Create a function called `create_student_summary` with 5 parameters:

1. `name`
2. `math_score`
3. `english_score`
4. `science_score`
5. `attendance`

Return a dictionary with the student's name, average score, and attendance.

````python
# Your answer for function exercise 3
def student_report(name, math_score, english_score, science_score, attendance):
    average_score = (math_score + english_score + science_score) / 3

    return {
        "Name": name,
        "Average Scpre": average_score,
        "Attendance": attendance
    }

print(student_report("Adam", math_score=95, english_score=80, science_score=95, attendance="Excellent"))
````

## Part 4: Class Exercises

A class is a blueprint for creating objects. Objects can store data and use methods.

### Exercise 10: Student Class

Create a class called `Student` with 3 parameters:

1. `name`
2. `age`
3. `course`

Add a method called `introduce` that returns a sentence about the student.

````python
# Your answer for class exercise 1
class Student:
    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course

    def introduce(self):
        return f"Hi, my name is {self.name}, I am {self.age} years old and I am currently studying {self.course}"
    
student = Student("Adam", 35, "Python")
print(student.introduce())
````

### Exercise 11: Bank Account Class

Create a class called `BankAccount` with 4 parameters:

1. `owner`
2. `bank_name`
3. `account_type`
4. `balance`

Add two methods:

1. `deposit(amount)`
2. `withdraw(amount)`

````python
# Your answer for class exercise 2
class BankAccount:
    def __init__(self, owner, bank_name, account_type, balance):
        self.owner = owner
        self.bank_name = bank_name
        self.account_type = account_type
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount
        return self.balance
    
    def withdraw(self, amount):
        if amount < balance:
            return "Not Enough Money"
        self.balance = self.balance - amount
        return self.balance
    
account = BankAccount("Adam", "BofA", "Savings", 1000)
print(account.deposit(200))
print(account.withdraw(600))

    
````
