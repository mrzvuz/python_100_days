# Connecting Error Handling With Loops, Functions, and Classes

This notebook connects several Python ideas together:

1. error handling with `try` and `except`
2. `for` loops
3. `while` loops
4. functions with 4 parameters
5. classes with 4 parameters

The goal is to see how professional Python code often combines these pieces instead of using them separately.

## Big Idea

Error handling is not separate from loops, functions, and classes.

In real programs:

- loops process many values
- functions organize repeated logic
- classes store data and behavior
- error handling protects risky operations

You usually combine them.

## 1. Error Handling Inside a For Loop

A `for` loop often processes many items.

Sometimes one item is bad, but you do not want the whole loop to crash.

````python
number_texts = ["10", "25", "hello", "40", "abc", "5"]

valid_numbers = []

for text in number_texts:
    try:
        number = int(text)
        valid_numbers.append(number)
    except ValueError:
        print("Skipping invalid number:", text)

print("Valid numbers:", valid_numbers)
print("Total:", sum(valid_numbers))
````

What happened?

1. The loop checks each value.
2. `int(text)` is risky because some text cannot become a number.
3. If conversion works, the number is added to `valid_numbers`.
4. If conversion fails, `except ValueError` handles it.
5. The loop continues instead of crashing.

## 2. Error Handling Inside a While Loop

A `while` loop is useful when you want to keep trying until something works.

In this example, we simulate user input using a list, so the notebook can run without waiting for you to type.

````python
attempts = ["abc", "-5", "30"]
index = 0
age = None

while age is None and index < len(attempts):
    text = attempts[index]
    index = index + 1

    try:
        possible_age = int(text)

        if possible_age < 0:
            raise ValueError("Age cannot be negative")

        age = possible_age
    except ValueError as error:
        print("Invalid age:", text, "|", error)

print("Final age:", age)
````

Notice two things:

- `int(text)` can raise a `ValueError` if the text is not numeric.
- We can also use `raise ValueError(...)` ourselves when the number is invalid for our program.

Professional code often validates data early like this.

## 3. Function With 4 Parameters and Error Handling

This function has 4 parameters:

1. `product_name`
2. `quantity_text`
3. `price_text`
4. `tax_rate_text`

The function receives text values and safely converts them to numbers.

````python
def calculate_order_total(product_name, quantity_text, price_text, tax_rate_text):
    try:
        quantity = int(quantity_text)
        price = float(price_text)
        tax_rate = float(tax_rate_text)

        if quantity < 0:
            raise ValueError("quantity cannot be negative")

        if price < 0:
            raise ValueError("price cannot be negative")

        subtotal = quantity * price
        total = subtotal + subtotal * tax_rate
        return {
            "product": product_name,
            "total": total,
            "success": True,
        }

    except ValueError as error:
        return {
            "product": product_name,
            "error": str(error),
            "success": False,
        }


print(calculate_order_total("Notebook", "3", "15", "0.20"))
print(calculate_order_total("Pen", "ten", "2", "0.20"))
print(calculate_order_total("Laptop", "1", "-1200", "0.20"))
````

Why this is professional:

- the risky conversion code is inside `try`
- specific errors are handled with `except ValueError`
- invalid business rules raise clear errors
- the function returns structured data instead of crashing

## 4. For Loop + Function + Error Handling

Now we use a `for` loop to call the function many times.

Some orders are valid. Some are invalid.

````python
orders = [
    ["Notebook", "3", "15", "0.20"],
    ["Pen", "ten", "2", "0.20"],
    ["Laptop", "1", "1200", "0.20"],
    ["Mouse", "4", "abc", "0.20"],
]

for order in orders:
    result = calculate_order_total(
        order[0],
        order[1],
        order[2],
        order[3],
    )

    if result["success"]:
        print(result["product"], "total:", result["total"])
    else:
        print(result["product"], "failed:", result["error"])
````

This is a very important pattern:

`for loop` processes many records.

`function` handles one record.

`try/except` protects risky conversion and validation.

## 5. Class With 4 Parameters and Error Handling

This class has 4 parameters:

1. `owner`
2. `bank_name`
3. `account_type`
4. `balance`

The class validates the balance and protects deposit/withdraw operations.

````python
class BankAccount:
    def __init__(self, owner, bank_name, account_type, balance):
        if balance < 0:
            raise ValueError("Starting balance cannot be negative")

        self.owner = owner
        self.bank_name = bank_name
        self.account_type = account_type
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive")

        self.balance = self.balance + amount
        return self.balance

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive")

        if amount > self.balance:
            raise ValueError("Not enough money")

        self.balance = self.balance - amount
        return self.balance

    def summary(self):
        return f"{self.owner} has {self.balance} in a {self.account_type} account at {self.bank_name}."
````

The class itself raises errors when someone tries to use it incorrectly.

Then the code that uses the class can decide how to handle those errors.

````python
try:
    account = BankAccount("Adam", "Python Bank", "Savings", 1000)
    print(account.summary())
    print(account.deposit(250))
    print(account.withdraw(2000))
except ValueError as error:
    print("Bank account error:", error)
````

## 6. While Loop + Class + Error Handling

Now we simulate multiple bank operations.

A `while` loop keeps processing until there are no more operations.

````python
account = BankAccount("Sara", "Python Bank", "Checking", 500)

operations = [
    ["deposit", 100],
    ["withdraw", 50],
    ["withdraw", 1000],
    ["deposit", -20],
]

index = 0

while index < len(operations):
    operation = operations[index]
    index = index + 1

    action = operation[0]
    amount = operation[1]

    try:
        if action == "deposit":
            new_balance = account.deposit(amount)
        elif action == "withdraw":
            new_balance = account.withdraw(amount)
        else:
            raise ValueError("Unknown operation")

        print(action, amount, "worked. Balance:", new_balance)

    except ValueError as error:
        print(action, amount, "failed:", error)

print(account.summary())
````

## Professional Best Practices When Combining These Pieces

1. Keep risky code inside a small `try` block.
2. Catch specific errors like `ValueError`, `KeyError`, or `FileNotFoundError`.
3. Do not hide errors with empty `except` blocks.
4. Validate data near the place where it matters.
5. Use functions to handle one item clearly.
6. Use loops to repeat that function over many items.
7. Use classes when you need objects that remember state over time.
8. Raise clear errors inside classes when someone uses the object incorrectly.

# Possible Solutions

## Solution 1: For Loop + Error Handling

````python
scores = ["90", "85", "bad", "100", "missing"]
valid_scores = []

for score_text in scores:
    try:
        score = int(score_text)
        valid_scores.append(score)
    except ValueError:
        print("Skipping invalid score:", score_text)

print("Valid scores:", valid_scores)
print("Average:", sum(valid_scores) / len(valid_scores))
````

## Solution 2: Function With 4 Parameters + Error Handling

````python
def calculate_salary(employee_name, hours_text, rate_text, bonus_text):
    try:
        hours = float(hours_text)
        rate = float(rate_text)
        bonus = float(bonus_text)

        if hours < 0:
            raise ValueError("hours cannot be negative")

        if rate < 0:
            raise ValueError("rate cannot be negative")

        salary = hours * rate + bonus
        return {
            "success": True,
            "employee": employee_name,
            "salary": salary,
        }

    except ValueError as error:
        return {
            "success": False,
            "employee": employee_name,
            "error": str(error),
        }


print(calculate_salary("Adam", "40", "25", "200"))
print(calculate_salary("Sara", "forty", "25", "200"))
````

## Solution 3: Class With 4 Parameters + Error Handling

````python
class Product:
    def __init__(self, name, category, price, stock):
        if price < 0:
            raise ValueError("price cannot be negative")

        if stock < 0:
            raise ValueError("stock cannot be negative")

        self.name = name
        self.category = category
        self.price = price
        self.stock = stock

    def total_value(self):
        return self.price * self.stock


product_data = [
    ["Laptop", "Electronics", 1200, 5],
    ["Mouse", "Electronics", 25, 50],
    ["Broken Item", "Unknown", -10, 3],
    ["Notebook", "Stationery", 4, 100],
]

products = []

for item in product_data:
    try:
        product = Product(item[0], item[1], item[2], item[3])
        products.append(product)
        print(product.name, "value:", product.total_value())
    except ValueError as error:
        print(item[0], "failed:", error)
````

## Quick Memory Guide

- Use a `for` loop when you have many items to process.
- Use a `while` loop when you need to keep going until a condition changes.
- Use a function when you want reusable logic.
- Use a class when you need objects that remember their own data.
- Use `try`/`except` around code that might fail.
- Use `raise` when your own code detects invalid data.

The professional pattern is often:

`loop -> function or class method -> validation -> error handling -> continue safely`
