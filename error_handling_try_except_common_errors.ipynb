# Error Handling, Try/Except, and Common Python Errors

Errors are normal in programming. Professional Python developers do not try to avoid every possible error by guessing. Instead, they write code that is clear, testable, and handles expected problems carefully.

In this notebook, you will learn:

1. what errors and exceptions are
2. common Python errors
3. how `try` and `except` work
4. professional best practices
5. beginner-friendly practice exercises with solutions

## 1. What Is an Error?

An error means Python found a problem while reading or running your code.

There are two broad kinds of problems:

1. **Syntax errors**: Python cannot understand the code.
2. **Exceptions**: Python understands the code, but something goes wrong while running it.

## 2. Syntax Error

A syntax error happens when Python cannot read your code because the grammar is wrong.

Example with a missing colon:

```python
if age > 18
    print("Adult")
```

Correct version:

```python
if age > 18:
    print("Adult")
```

## 3. Exceptions

An exception happens while the program is running.

For example, this code is valid Python, but it fails when it runs because you cannot divide by zero.

````python
# Uncomment this line to see the error:
# print(10 / 0)
````

The error would be:

```text
ZeroDivisionError: division by zero
```

The name `ZeroDivisionError` tells you what kind of problem happened.

## 4. Common Python Errors

Here are common errors beginners see often.

### NameError

A `NameError` happens when you use a variable name that does not exist.

````python
# Uncomment to see NameError:
# print(username)

username = "Adam"
print(username)
````

### TypeError

A `TypeError` happens when you use a value in a way that does not match its type.

````python
# Uncomment to see TypeError:
# print("Age: " + 25)

print("Age: " + str(25))
````

### ValueError

A `ValueError` happens when the type is correct, but the value is invalid.

````python
# Uncomment to see ValueError:
# number = int("hello")

number = int("42")
print(number)
````

### IndexError

An `IndexError` happens when you try to access a list position that does not exist.

````python
colors = ["red", "green", "blue"]

# Uncomment to see IndexError:
# print(colors[5])

print(colors[0])
````

### KeyError

A `KeyError` happens when you try to access a dictionary key that does not exist.

````python
student = {"name": "Sara", "age": 22}

# Uncomment to see KeyError:
# print(student["grade"])

print(student.get("grade", "No grade yet"))
````

### FileNotFoundError

A `FileNotFoundError` happens when Python tries to open a file that does not exist.

````python
from pathlib import Path

path = Path("missing_file.txt")

if path.exists():
    print(path.read_text(encoding="utf-8"))
else:
    print("File does not exist yet.")
````

## 5. Basic Try/Except

`try` means: run this code.

`except` means: if a specific error happens, run this backup code.

````python
try:
    number = int("hello")
    print(number)
except ValueError:
    print("That value cannot be converted to an integer.")
````

The program does not crash. Instead, Python catches the `ValueError` and prints a friendly message.

## 6. Catching the Error Object

You can store the error message in a variable. Professionals often use `as error` or `as exc`.

````python
try:
    result = 10 / 0
except ZeroDivisionError as error:
    print("Something went wrong:", error)
````

## 7. Multiple Except Blocks

Different errors should usually be handled differently.

````python
value = "abc"
divider = 0

try:
    number = int(value)
    result = number / divider
    print(result)
except ValueError:
    print("Please provide a valid number.")
except ZeroDivisionError:
    print("Cannot divide by zero.")
````

## 8. Else and Finally

`else` runs only if no exception happened.

`finally` always runs, whether an exception happened or not.

````python
try:
    number = int("42")
except ValueError:
    print("Invalid number")
else:
    print("Conversion worked:", number)
finally:
    print("Finished trying to convert the value.")
````

## 9. Raising Your Own Errors

Sometimes your function should reject invalid input. Use `raise` to create a clear error.

````python
def calculate_discount(price, discount_percent):
    if price < 0:
        raise ValueError("price cannot be negative")

    if discount_percent < 0 or discount_percent > 100:
        raise ValueError("discount_percent must be between 0 and 100")

    discount = price * discount_percent / 100
    return price - discount


print(calculate_discount(100, 20))
````

## 10. Professional Best Practices

### 1. Catch specific exceptions

Prefer this:

```python
except ValueError:
```

Avoid this when possible:

```python
except Exception:
```

Specific exceptions make your code easier to debug.

### 2. Keep the `try` block small

Put only the risky line or risky operation inside `try`.

### 3. Do not hide errors silently

Avoid this:

```python
except ValueError:
    pass
```

If something goes wrong, handle it or explain it.

### 4. Use clear error messages

Good error messages tell the user or developer what went wrong.

### 5. Validate data early

If a function needs a positive number, check that at the beginning.

### 6. Use `with` for files

`with` automatically closes files, even if an error happens.

### 7. Let unexpected errors be visible

Do not catch every error just to make the program look like it works. Unexpected errors often reveal real bugs.

## 11. Example: Safe Number Conversion

This function tries to convert text into an integer. If it cannot, it returns `None`.

````python
def safe_int(value):
    try:
        return int(value)
    except ValueError:
        return None


print(safe_int("42"))
print(safe_int("hello"))
````

## 12. Example: Safe Dictionary Access

For dictionaries, `.get()` is often cleaner than `try`/`except` when missing keys are expected.

````python
student = {"name": "Sara", "age": 22}

grade = student.get("grade", "No grade yet")
print(grade)
````

## 13. Example: Safe File Reading

File reading can fail if the file does not exist. Handle that expected case clearly.

````python
from pathlib import Path

def read_file_safely(file_path):
    path = Path(file_path)

    try:
        return path.read_text(encoding="utf-8")
    except FileNotFoundError:
        return "File was not found."


print(read_file_safely("missing_file.txt"))
````

# Possible Solutions

## Solution 1: Safe Division

````python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Cannot divide by zero"


print(safe_divide(10, 2))
print(safe_divide(10, 0))
````

## Solution 2: Convert Age

````python
def convert_age(age_text):
    try:
        return int(age_text)
    except ValueError:
        return "Invalid age"


print(convert_age("25"))
print(convert_age("twenty five"))
````

## Solution 3: Get Student Grade

````python
def get_student_grade(student, key):
    try:
        return student[key]
    except KeyError:
        return "Grade not found"


student = {"name": "Sara", "math": 90, "english": 85}

print(get_student_grade(student, "math"))
print(get_student_grade(student, "science"))
````

## Quick Memory Guide

- `try`: code that might fail
- `except`: code that runs if a specific error happens
- `else`: code that runs if no error happens
- `finally`: code that always runs
- `raise`: create your own error

Professional habit:

Catch the errors you expect. Let unexpected errors teach you what needs fixing.
